```python
import json
import time
import random
from datetime import datetime
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers='localhost:9092',
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

accounts = [('ICIC0000072', 123456), ('ICIC0000072', 123457), ('SBIN0004426', 200001), ('SBIN0004426', 200002), ('CNRB0010233', 400001), ('CNRB0010233', 400002) ]
transaction = ['C', 'D', 'F']


def generate_transaction_event():
    account_detail = random.choice(accounts)
    return {
        "txn_id": account_detail[1],
        "IFSC_Code": account_detail[0],
        "txn_type": random.choice(transaction),
        "timestamp": datetime.now().isoformat(),
        "amount": random.randint(1, 200000)
    }

print("Producing Transaction events...")

while True:
    event = generate_transaction_event()
    producer.send("transactions-count1", event)
    print("Sent:", event)
    time.sleep(1)
```
