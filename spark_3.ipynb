```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark=SparkSession.builder.appName("Weather Data Analysis").getOrCreate()
```

```python
filepath='D:\BDA_0024\ABD_LAB\datasets\Weather Data in India from 1901 to 2017.csv'
```

```python
df = spark.read.csv(filepath,header=True,inferSchema=True)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df = df.withColumn(
    'Avg_Temp',
    round(
        (col('JAN') + col('FEB') + col('MAR') + col('APR') +
         col('MAY') + col('JUN') + col('JUL') + col('AUG') +
         col('SEP') + col('OCT') + col('NOV') + col('DEC')) / 12,
        2
    )
)

df.select('YEAR', 'Avg_Temp').show()
```

```python
# 2. Add row average temperature of each month 
df.select([
    round(avg(col(month)), 2).alias(month)
    for month in months
]).show()
```

```python
# 3. Add columns Min and Max for each year with corresponding values 
df = df.withColumn(
    'Min_Temp',
    least(*[col(month) for month in months])
)

df = df.withColumn(
    'Max_Temp',
    greatest(*[col(month) for month in months])
)

df.select('YEAR', 'Min_Temp', 'Max_Temp').show()
```

```python
# 4. Create dataframe called 'decade', storing average temperatures of each month in the respective decade. Decade 1910 is from 1901 to 1910. 
decade_df = df.withColumn(
    'Decade',
    ((col('YEAR') - 1) / 10).cast('int') * 10 + 10
)
```

```python
decade = decade_df.groupBy('Decade').agg(
    *[
        round(avg(col(month)), 2).alias(month)
        for month in months
    ]
).orderBy('Decade')

decade.show()
```

```python
# 5. Which year is the hottest year? (Based on avg temp) 
df.select('YEAR', 'Avg_Temp') \
    .orderBy(desc('Avg_Temp')) \
    .limit(1) \
    .show()
```

```python
# 6. Which year is the coldest year? (Based on avg temp)
df.select('YEAR', 'Avg_Temp') \
    .orderBy('Avg_Temp') \
    .limit(1) \
    .show()
```

```python
# 7. Find out which years recorded minimum and maximum temperatures. (Not based on avg temp) 
df.select(
    'YEAR',
    'Min_Temp'
).orderBy('Min_Temp').limit(1).show()
```

```python
df.select(
    'YEAR',
    'Max_Temp'
).orderBy(desc('Max_Temp')).limit(1).show()
```

```python
# 8. How much raise is observed in min and max temperature for each month?
monthly_rise = df.select([
    round(
        max(col(month)) - min(col(month)),
        2
    ).alias(month)
    for month in months
])

monthly_rise.show()
```
