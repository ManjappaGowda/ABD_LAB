```python
import json
import random
import time
from datetime import datetime, timezone

from kafka import KafkaProducer
```

```python
TOPIC = "food-orders"

producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda event: json.dumps(event).encode("utf-8")
)

print("Connected to Kafka.")
```

```python
restaurants = [
    {"restaurant_id": 101, "area": "Indiranagar", "cuisine": "Indian"},
    {"restaurant_id": 102, "area": "Whitefield", "cuisine": "Chinese"},
    {"restaurant_id": 103, "area": "Koramangala", "cuisine": "Italian"},
    {"restaurant_id": 104, "area": "Indiranagar", "cuisine": "Chinese"},
    {"restaurant_id": 105, "area": "Whitefield", "cuisine": "Indian"},
    {"restaurant_id": 106, "area": "Koramangala", "cuisine": "Indian"}
]

statuses = [
    "placed",
    "preparing",
    "picked_up",
    "delivered",
    "delayed",
    "cancelled"
]

order_id = 1000
```

```python
def generate_order():
    restaurant = random.choice(restaurants)

    status = random.choices(
        statuses,
        weights=[15, 20, 15, 30, 12, 8],
        k=1
    )[0]

    return {
        "order_id": order_id,
        "restaurant_id": restaurant["restaurant_id"],
        "delivery_area": restaurant["area"],
        "cuisine": restaurant["cuisine"],
        "status": status,
        "estimated_delivery_time": 30,
        "customer": f"Customer_{random.randint(1, 100)}",
        "event_time": datetime.now(timezone.utc).isoformat()
    }
```

```python
try:
    while True:
        event = generate_order()

        producer.send(TOPIC, event).get(timeout=10)
        print("Sent:", event)

        order_id += 1
        time.sleep(2)

except KeyboardInterrupt:
    print("Producer stopped.")

finally:
    producer.close()
```
